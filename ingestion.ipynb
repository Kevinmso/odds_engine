```python
gols_vasco_mandante = 9
gols_vasco_visitante = 7

print(gols_vasco_mandante)
print(gols_vasco_visitante)
```

```python
gols_sao_paulo_mandante = 10
gols_sao_paulo_visitante = 5

print(gols_sao_paulo_mandante)
print(gols_sao_paulo_visitante)
```

```python
gols_contra_sao_paulo_mandante = 3
gols_contra_sao_paulo_visitante = 6

print(gols_contra_sao_paulo_mandante)
print(gols_contra_sao_paulo_visitante)
```

```python
gols_contra_vasco_mandante = 7
gols_contra_vasco_visitante = 10

print(gols_contra_vasco_mandante)
print(gols_contra_vasco_visitante)
```

```python
# Médias separadas por contexto (mandante/visitante)
# A liga tem médias diferentes em casa e fora — usar a mesma distorce o cálculo
media_gols_mandante_liga = 1.4   # média de gols do mandante por jogo na liga
media_gols_visitante_liga = 1.1  # média de gols do visitante por jogo na liga

# Vasco joga como mandante nesta partida → usamos os dados de quando ele joga em casa
media_marcados_vasco   = gols_vasco_mandante           / 6
media_sofridos_vasco   = gols_contra_vasco_mandante    / 6

# São Paulo joga como visitante → usamos os dados de quando ele joga fora
media_marcados_sao_paulo  = gols_sao_paulo_visitante       / 5
media_sofridos_sao_paulo  = gols_contra_sao_paulo_visitante / 5

print(media_marcados_vasco)
print(media_sofridos_vasco)
print(media_marcados_sao_paulo)
print(media_sofridos_sao_paulo)
```

```python
from core.calculator.lmbda import calcTeamFactor, calcLambda

# Força de ataque normalizada pela média do contexto correspondente
forca_ataque_vasco      = calcTeamFactor(media_marcados_vasco, media_gols_mandante_liga)
forca_ataque_sao_paulo  = calcTeamFactor(media_marcados_sao_paulo, media_gols_visitante_liga)

print(forca_ataque_vasco)
print(forca_ataque_sao_paulo)
```

```python
# Força de defesa: normalizada pela média de quem ataca naquele contexto
# Vasco defende em casa → recebe visitantes → normaliza por media_gols_visitante_liga
forca_defesa_vasco     = calcTeamFactor(media_sofridos_vasco, media_gols_visitante_liga)

# São Paulo defende fora → recebe mandante → normaliza por media_gols_mandante_liga
forca_defesa_sao_paulo  = calcTeamFactor(media_sofridos_sao_paulo, media_gols_mandante_liga)

print(forca_defesa_vasco)
print(forca_defesa_sao_paulo)
```

```python
from core.model.poisson import PoissonModel

# lambda_home = ataque_mandante × defesa_visitante × media_mandante_liga
# lambda_away = ataque_visitante × defesa_mandante × media_visitante_liga
lambda_vasco = calcLambda(forca_ataque_vasco, forca_defesa_sao_paulo, media_gols_mandante_liga)
lambda_sao_paulo = calcLambda(forca_ataque_sao_paulo, forca_defesa_vasco, media_gols_visitante_liga)

model = PoissonModel(15)
probabilidades = model.matchOutcome(lambda_home=lambda_vasco, lambda_away=lambda_sao_paulo)
odds = model.getMatchOdds(lambda_home=lambda_vasco, lambda_away=lambda_sao_paulo)

print(f"PROBABILIDADES: {probabilidades}")
print(f"ODDS: {odds}")
```
