```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("data/campeonato-brasileiro-full.csv")
display(df)
```

```python
# converter campo data da base para o formato datetime
df["data"] = pd.to_datetime(df["data"], format="%d/%m/%Y")

df2025 = df[df["data"].dt.year == 2025]
display(df2025)
```
